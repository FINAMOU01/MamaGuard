```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

n = 600
data = []

for _ in range(n):
    
    risque = random.choices(
        ['Normale', 'Modere', 'Eleve'],
        weights=[60, 25, 15]
    )[0]

    if risque == 'Normale':
        # Valeurs normales AVEC légère variation
        bpm               = random.randint(60, 105)  # chevauchement
        temperature       = round(random.uniform(36.1, 37.8), 1)
        spo2              = random.randint(94, 100)  # chevauchement
        tension_s         = random.randint(100, 135) # chevauchement
        tension_d         = random.randint(65, 88)   # chevauchement
        contractions      = random.randint(0, 2)
        semaine           = random.randint(8, 40)

    elif risque == 'Modere':
        # Valeurs modérées — chevauchent avec Normale ET Elevé
        bpm               = random.randint(95, 125)  # chevauchement
        temperature       = round(random.uniform(37.4, 38.7), 1)
        spo2              = random.randint(91, 96)   # chevauchement
        tension_s         = random.randint(125, 148) # chevauchement
        tension_d         = random.randint(82, 95)   # chevauchement
        contractions      = random.randint(2, 5)
        semaine           = random.randint(8, 40)

    else:  # Eleve
        # Valeurs élevées AVEC légère variation
        bpm               = random.randint(115, 150) # chevauchement
        temperature       = round(random.uniform(38.3, 40.5), 1)
        spo2              = random.randint(85, 93)   # chevauchement
        tension_s         = random.randint(138, 180) # chevauchement
        tension_d         = random.randint(88, 120)  # chevauchement
        contractions      = random.randint(4, 10)
        semaine           = random.randint(8, 40)

    # Ajouter du bruit réaliste (cas ambigus)
    # 10% des cas ont un paramètre qui "trompe"
    if random.random() < 0.10:
        bpm = bpm + random.randint(-15, 15)
        bpm = max(50, min(160, bpm))  # garder dans limites

    data.append({
        'bpm'                    : bpm,
        'temperature'            : temperature,
        'spo2'                   : spo2,
        'tension_systolique'     : tension_s,
        'tension_diastolique'    : tension_d,
        'contractions_par_10min' : contractions,
        'semaine_grossesse'      : semaine,
        'risque'                 : risque
    })

df = pd.DataFrame(data)

# Ajouter le Feature Engineering directement
df['pulse_pressure'] = df['tension_systolique'] - df['tension_diastolique']
df['indice_tension']  = (df['tension_systolique'] / df['tension_diastolique']).round(2)
df['fievre']          = (df['temperature'] >= 37.6).astype(int)
df['spo2_basse']      = (df['spo2'] < 95).astype(int)

def get_trimestre(s):
    if s <= 12: return 1
    elif s <= 26: return 2
    else: return 3

df['trimestre'] = df['semaine_grossesse'].apply(get_trimestre)

df.to_csv('dataset_mamaguard1.csv', index=False)

print("=" * 45)
print("   DATASET MAMAGUARD V2 — GENERE")
print("=" * 45)
print(f"\n Nombre de cas     : {len(df)}")
print(f" Nombre de colonnes : {len(df.columns)}")
print(f"\n Repartition :")
print(df['risque'].value_counts())
print(f"\n Apercu :")
print(df.head(10))
```
